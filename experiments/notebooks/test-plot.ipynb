```python
import numpy as np
import matplotlib.pyplot as plt

def plot_C(C):
    C = np.array(C)

    plt.figure(figsize=(5,5))

    # origin
    plt.scatter(0, 0)
    plt.text(0, 0, " q")

    # plot each row as a vector
    for i, row in enumerate(C):
        plt.quiver(
            0, 0,
            row[0], row[1],
            angles="xy", scale_units="xy", scale=1
        )
        plt.text(row[0], row[1], f"row {i+1}")

    plt.axhline(0)
    plt.axvline(0)
    plt.gca().set_aspect("equal", adjustable="box")
    plt.xlim(-2, 2)
    plt.ylim(-2, 2)
    plt.xlabel("theta")
    plt.ylabel("dz")
    plt.title("Rows of C (gradients)")
    plt.grid()
    plt.show()
```

```python
C = [[-0.2, 1.0]]
plot_C(C)
```
